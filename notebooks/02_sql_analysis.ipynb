```python
import pandas as pd
from sqlalchemy import create_engine, text

engine = create_engine('postgresql://postgres:0987654321@localhost:5432/walmart_sales')

# Load cleaned data
df = pd.read_csv('../data/cleaned_data.csv')

# Push to PostgreSQL
df.to_sql('walmart_sales', engine, if_exists='replace', index=False)

print("Data loaded into PostgreSQL successfully")
print("Total rows:", len(df))
```

```python
query1 = """
    SELECT "Store", 
           ROUND(SUM("Weekly_Sales")::numeric, 2) AS total_sales
    FROM walmart_sales
    GROUP BY "Store"
    ORDER BY total_sales DESC
    LIMIT 10;
"""

df_q1 = pd.read_sql(query1, engine)
print("Top 10 Stores by Total Sales:")
print(df_q1)
```

```python
query2 = """
    SELECT "IsHoliday",
           COUNT(*) AS weeks,
           ROUND(AVG("Weekly_Sales")::numeric, 2) AS avg_sales,
           ROUND(SUM("Weekly_Sales")::numeric, 2) AS total_sales
    FROM walmart_sales
    GROUP BY "IsHoliday";
"""

df_q2 = pd.read_sql(query2, engine)
print("Holiday vs Non-Holiday Sales:")
print(df_q2)
```

```python
query3 = """
    SELECT 
        DATE_TRUNC('month', "Date"::timestamp) AS month,
        ROUND(SUM("Weekly_Sales")::numeric, 2) AS monthly_sales
    FROM walmart_sales
    GROUP BY month
    ORDER BY month;
"""

df_q3 = pd.read_sql(query3, engine)
print("Monthly Sales Trend:")
print(df_q3)
```

```python
query4 = """
    SELECT "Dept",
           ROUND(AVG("Weekly_Sales")::numeric, 2) AS avg_weekly_sales,
           ROUND(SUM("Weekly_Sales")::numeric, 2) AS total_sales
    FROM walmart_sales
    WHERE "Weekly_Sales" > 0
    GROUP BY "Dept"
    ORDER BY avg_weekly_sales DESC
    LIMIT 5;
"""

df_q4 = pd.read_sql(query4, engine)
print("Top 5 Departments by Avg Weekly Sales:")
print(df_q4)
```

```python
# Save all SQL results for Power BI
df_q1.to_csv('../outputs/top_stores.csv', index=False)
df_q2.to_csv('../outputs/holiday_impact.csv', index=False)
df_q3.to_csv('../outputs/monthly_trend.csv', index=False)
df_q4.to_csv('../outputs/top_departments.csv', index=False)

print("All SQL results saved to outputs folder")
```
